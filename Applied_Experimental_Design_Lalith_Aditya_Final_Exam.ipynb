4. Why might a full factorial design be preferred here to a $2^k$ fractional factorial design?

**Ans**:

A **full factorial design** was selected rather than a fractional \(2^k\) design for the following reasons:

- **No aliasing (all effects are estimable):**  
  With a \(3 \times 2 \times 2\) design and **5 replications per treatment combination**, we can estimate:
  - All **main effects**
  - All **two-way interactions**
  - The **three-way interaction**  
  independently, without confounding.

- **The design is still small and feasible:**  
  The total number of observations is:
  \[
  3 \times 2 \times 2 \times 5 = 60
  \]
  This is practical to run and does not require excessive time or cost.

- **Interactions matter for food preference studies:**  
  In taste experiments, it’s common for the effect of one ingredient to depend on another (for example, a specific protein might pair better with a particular salsa level). A fractional factorial design could **confound** important interactions, increasing the chance of missing meaningful ingredient combinations.

5. State all sets of statistical hypotheses.

Let \(\mu_{ijk}\) denote the **mean satisfaction score** for:

- Protein level \(i\)  
- Salsa level \(j\)  
- Potato level \(k\)



### Main Effects Hypotheses

#### **Protein**
- \(H_0:\ \mu_{\text{Bacon}} = \mu_{\text{Sausage}} = \mu_{\text{Chorizo}}\)  
- \(H_A:\) At least one protein mean differs

#### **Salsa**
- \(H_0:\ \mu_{\text{Mild}} = \mu_{\text{Hot}}\)  
- \(H_A:\ \mu_{\text{Mild}} \ne \mu_{\text{Hot}}\)

#### **Potatoes**
- \(H_0:\ \mu_{\text{Yes}} = \mu_{\text{No}}\)  
- \(H_A:\ \mu_{\text{Yes}} \ne \mu_{\text{No}}\)



### Interaction Effects Hypotheses

Each interaction tests whether the effect of one factor **depends on** the level of another factor:

- Protein \(\times\) Salsa  
- Protein \(\times\) Potatoes  
- Salsa \(\times\) Potatoes  
- Protein \(\times\) Salsa \(\times\) Potatoes  

**General form (for any interaction):**
- \(H_0:\) No interaction effect  
- \(H_A:\) An interaction exists

7. Fit the full factorial ANOVA model. Check the assumption of normality using a testing method and a visual method. Does the assumption hold? Why or why not?

**Ans**:

A full factorial ANOVA was fitted, meaning the model included every main ingredient effect (Protein, Salsa, and Potatoes) along with all possible interaction terms. The results suggest that Protein and Salsa both matter for average satisfaction: Protein was statistically significant (p = 0.0247), and Salsa was also significant, though only just (p = 0.0495). Potatoes, however, did not meet the usual 5% significance cutoff (p = 0.0868). That doesn’t mean potatoes have no effect—it just means the evidence isn’t strong enough here to confidently conclude an effect at the standard threshold.

Notably, none of the interaction terms were significant, which indicates there isn’t strong evidence that the impact of one ingredient changes depending on another ingredient. In practical terms, satisfaction seems to be driven mostly by separate (“additive”) effects of ingredients rather than by specific ingredient combinations creating extra boosts or drops in satisfaction.

Finally, the ANOVA normality assumption was checked using a Shapiro–Wilk test and a Q–Q plot. Since the Shapiro–Wilk test was not significant (p = 0.135) and the Q–Q plot showed points roughly lining up along a straight line, the residuals look approximately normal. That supports the ANOVA assumptions, so the results above are reasonable to interpret.

```python
## Question 7 Code ##

import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.formula.api import ols

# -------------------------------------
# Load data
# -------------------------------------
df = pd.read_excel("Busters Burritos.xlsx")

# -------------------------------------
# Full Factorial ANOVA Model
# -------------------------------------
# Includes all main effects and interactions
model = ols(
    "Satisfaction ~ C(Protein) * C(Salsa) * C(Potatoes)",
    data=df
).fit()

anova_table = sm.stats.anova_lm(model, typ=2)

print("\n=== FULL FACTORIAL ANOVA TABLE ===")
print(anova_table)

# -------------------------------------
# Normality Testing: Shapiro-Wilk Test
# -------------------------------------
residuals = model.resid

shapiro_test = stats.shapiro(residuals)

print("\n=== SHAPIRO-WILK NORMALITY TEST ===")
print("Test Statistic =", shapiro_test.statistic)
print("p-value =", shapiro_test.pvalue)

# -------------------------------------
# Visual Normality Check: Q-Q Plot
# -------------------------------------
plt.figure()
stats.probplot(residuals, plot=plt)
plt.title("Normal Q-Q Plot of Residuals")
plt.savefig("Buster's Burritos_QQ_Plot_Residuals.png", dpi=300)
plt.show()
```

8. Check the assumption of constant variance using a testing method and a visual method. Does the assumption hold? Why or why not?

**Ans**:

Constant variance (homoscedasticity) was checked in two ways: Levene’s test and a residuals-versus-fitted plot. Levene’s test was not significant (p > 0.05), so it did not provide evidence that the variances differ across the treatment combinations. The residual plot also supported this—residuals were scattered randomly around zero with a fairly even spread across the fitted values, and there was no clear funnel or widening/narrowing pattern. Taken together, these results suggest the equal-variance assumption for ANOVA is reasonable and appears to be satisfied.

```python
## Question 8 Code ##
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.formula.api import ols

# -------------------------------------
# Load data
# -------------------------------------
df = pd.read_excel("Busters Burritos.xlsx")

# -------------------------------------
# Fit Full Factorial ANOVA Model
# -------------------------------------
model = ols(
    "Satisfaction ~ C(Protein)*C(Salsa)*C(Potatoes)",
    data=df
).fit()

# -------------------------------------
# Extract residuals & fitted values
# -------------------------------------
residuals = model.resid
fitted = model.fittedvalues

# -------------------------------------
# FORMAL TEST: Levene's Test of Equal Variances
# -------------------------------------
groups = []

# Create groups for Levene using full treatment combinations
for name, g in df.groupby(["Protein","Salsa","Potatoes"]):
    groups.append(g["Satisfaction"])

lev_test = stats.levene(*groups)

print("\n=== LEVENE TEST FOR CONSTANT VARIANCE ===")
print("Test Statistic =", lev_test.statistic)
print("p-value =", lev_test.pvalue)

# -------------------------------------
# VISUAL CHECK: Residuals vs Fitted Plot
# -------------------------------------
plt.figure()
plt.scatter(fitted, residuals)
plt.axhline(0)
plt.title("Residuals vs Fitted Values")
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
plt.savefig("Buster's Burritos_Residuals_vs_Fitted.png", dpi=300)
plt.show()
```

9. Report the ANOVA table. What statistical conclusions can we draw about our main effects and interaction effects?

**Ans**:

The factorial ANOVA showed that Protein (p = 0.0247) and Salsa (p = 0.0495) both have a real influence on customer satisfaction, meaning people’s ratings change depending on the protein type and the salsa heat level. Potatoes did not meet the usual 5% significance cutoff (p = 0.0868), so there isn’t strong enough evidence here to say potatoes clearly affect satisfaction.

Also, none of the two-way or three-way interactions were significant, which suggests the ingredients don’t “work differently” in certain combinations. In other words, the effects appear mostly independent and additive. Overall, satisfaction seems to be driven mainly by protein choice and salsa heat, rather than by specific ingredient pairings or special combinations..

```python
## Question 9 Code ##
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# -------------------------------------
# Load Data
# -------------------------------------
df = pd.read_excel("Busters Burritos.xlsx")

# -------------------------------------
# Fit Full Factorial ANOVA Model
# -------------------------------------
model = ols(
    "Satisfaction ~ C(Protein) * C(Salsa) * C(Potatoes)",
    data=df
).fit()

# -------------------------------------
# Generate ANOVA Table (Type II)
# -------------------------------------
anova_table = sm.stats.anova_lm(model, typ=2)

# -------------------------------------
# Print Results
# -------------------------------------
print("\n=== FULL FACTORIAL ANOVA TABLE ===")
print(anova_table)

# -------------------------------------
# Interpretation Helper
# -------------------------------------
alpha = 0.05
print("\n=== SIGNIFICANCE SUMMARY (α = 0.05) ===")

for row in anova_table.index:
    if row != "Residual":
        pval = anova_table.loc[row, "PR(>F)"]
        signif = "SIGNIFICANT" if pval < alpha else "Not Significant"
        print(f"{row:40}  p = {pval:.4f}   --> {signif}")
```

10. If the interactions and/or main effects were found to be statistically meaningful, conduct appropriate post-hoc tests. Report the results of these tests and any conclusions you can draw from them. If the interactions and/or main effects were not found to be statistically meaningful, explain why post-hoc tests would not be appropriate.

**Ans**:

Since the factorial ANOVA showed significant main effects for Protein and Salsa, it made sense to follow up with post-hoc tests only where they were actually needed. Protein was the only significant factor with more than two levels, so it was the only one that required multiple-comparisons testing.

Using Tukey’s HSD, the results showed that chorizo burritos were rated significantly higher than sausage burritos (mean difference = −0.92, p = 0.023). In contrast, bacon was not significantly different from either chorizo or sausage (p > 0.12 for both comparisons), suggesting bacon sits in the middle without a clear separation from the other two.

No post-hoc test was necessary for Salsa because it has only two levels, and the ANOVA main effect already directly tests that single comparison. Post-hoc testing was also not done for Potatoes or any interaction terms because those effects were not significant. Running multiple comparisons after a non-significant omnibus test would be unjustified and would increase the chance of false positives (inflating Type I error).

```python
## Question 10 Code ##
from statsmodels.stats.multicomp import pairwise_tukeyhsd
import pandas as pd

# Load data
df = pd.read_excel("Busters Burritos.xlsx")

# Post-hoc: Tukey HSD for Protein
tukey = pairwise_tukeyhsd(
    endog=df["Satisfaction"],
    groups=df["Protein"],
    alpha=0.05
)

print("\n=== TUKEY HSD: PROTEIN COMPARISONS ===")
print(tukey)

# Optional visualization
tukey.plot_simultaneous()
plt.title("Tukey HSD Confidence Intervals – Protein Means")
plt.savefig("Buster's Burritos_Tukey_Protein.png", dpi=300)
plt.show()
```

11. Calculate and interpret partial $\eta^2$ for the main effects.

**Ans**:

The partial η² results help put the ANOVA findings into practical terms by showing how much variation each factor explains after accounting for the others. Protein had the strongest impact, with a large effect (η²p ≈ 0.143), meaning it explains about 14% of the residual-adjusted variance in satisfaction. Salsa showed a moderate effect (η²p ≈ 0.078), accounting for roughly 8% of the variance.

Interestingly, Potatoes also had a moderate effect size (η²p ≈ 0.060, about 6%), even though it wasn’t statistically significant in the ANOVA. This suggests the potato effect might be real but smaller, more variable, or harder to detect with the current sample size—so the study may not have had enough power to clearly confirm it.

Overall, these effect sizes reinforce that protein choice is the most practically important driver of satisfaction, with salsa playing a secondary but still meaningful role.

```python
## Question 11 Code ##

# Partial eta-squared calculation using ANOVA table values

SSE = 54.276    # Residual sum of squares from ANOVA

effects = {
    "Protein": 9.052,
    "Salsa": 4.592667,
    "Potatoes": 3.456
}

print("\n=== PARTIAL ETA-SQUARED (η²p) ===")

for effect, SS in effects.items():
    eta = SS / (SS + SSE)
    print(effect, "η²p =", round(eta, 4))
```

4. Explain why a fractional design is appropriate here. What are the advantages and disadvantages of using a fractional design?

**Ans**:

A **$2^{3-1}$ fractional factorial design** was chosen instead of the full **$2^3 = 8$** design to reduce experimental cost and the day-to-day operational load.

 Advantages

- **Fewer runs:** Requires only **4 treatment combinations** instead of 8, cutting the testing time in half.

Disadvantages

- **Aliasing (confounding):** Some effects are mixed together (for example, **main effects can be confounded with interaction effects**).  
- **Limited interaction detection:** Interactions cannot be estimated cleanly or separately.  
- **Lower resolution:** An effect may appear important simply because it is combined with another effect, making interpretation less certain.

7. Assuming a defining relation of $I=ABC$, specify the aliasing structure of the design. What does this mean for the interpretation of the main effects and interactions?

**Ans**:
This $2^{3-1}$ fractional factorial design uses the defining relation $I = ABC$, which makes it a Resolution III design. In a Resolution III design, each main effect is aliased (confounded) with a two-factor interaction, so main effects cannot be interpreted as “pure” effects on their own.

For example, the estimated effect of Location (A) is confounded with the Timing × Marketing interaction ($BC$). That means an observed “main effect” could reflect the true main effect, the aliased interaction, or some combination of both.

Because of this, interpretation usually depends on assuming that two-factor interactions are small/negligible. If that assumption is questionable, the clean fix is follow-up experimentation—for instance, running a foldover or moving to a full factorial design—to separate main effects from interactions.

```python
# Aliasing structure for 2^(3-1) fractional factorial
# Defining relation: I = ABC

def multiply(e1, e2):
    # symmetric difference implements factorial cancellation
    return ''.join(sorted(set(e1) ^ set(e2))) or "I"

effects = ["I","A","B","C","AB","AC","BC","ABC"]
generator = "ABC"

print("Defining relation: I = ABC\n")
print("Alias structure:")

for e in effects:
    alias = multiply(e.replace("I",""), generator)
    print(f"{e:>3}  ≡  {e} + {alias}")
```

8. Fit the appropriate ANOVA model to the data. Assess the normality assumption using two methods. Do we have evidence to support the assumption of normality? Why or why not?

**Ans**:

## ANOVA Model

An ANOVA model was fit using **Location**, **Timing**, and **Marketing** as main effects. Because this is a **$2^{3-1}$ fractional factorial design** with defining relation **$I = ABC$**, each main effect is **aliased with a two-factor interaction** (for example, **Location (A)** is aliased with **Timing × Marketing (BC)**).  

As a result, the estimated “main effects” should be interpreted as **combined effects** of the main factor **and** its aliased interaction. Any conclusions must be made with this confounding in mind.

## Normality Assessment

Residual normality was evaluated using two approaches:

- **Shapiro–Wilk test**
  - **$H_0$:** Residuals are normally distributed  
  - The test produced **$p > 0.05$**, so we **fail to reject $H_0$**.

- **Q–Q plot of residuals**
  - Residual points fall **approximately along the reference line**, with no strong curvature or skewness.
  - Minor deviations in the tails are present, but they are consistent with typical random variation.

## Conclusion on Normality

Overall, there is **support for the normality assumption**. Both the Shapiro–Wilk test and the Q–Q plot suggest the residuals are **approximately normal**, and with a balanced design and moderate sample size, the ANOVA results are generally **robust to small departures from normality**.

```python
## Question 8 Code ##
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.formula.api import ols

# -------------------------------------
# Load data
# -------------------------------------
df = pd.read_excel("Busters Burrito Sales.xlsx")

# -------------------------------------
# Fit ANOVA model (main effects only)
# -------------------------------------
model = ols(
    "Sales ~ C(Location) + C(Timing) + C(Marketing)",
    data=df
).fit()

anova_table = sm.stats.anova_lm(model, typ=2)

print("\n=== ANOVA TABLE (Fractional Factorial) ===")
print(anova_table)

# -------------------------------------
# Normality Check 1: Shapiro–Wilk Test
# -------------------------------------
residuals = model.resid
shapiro = stats.shapiro(residuals)

print("\n=== SHAPIRO–WILK TEST ===")
print("Test statistic:", shapiro.statistic)
print("p-value:", shapiro.pvalue)

# -------------------------------------
# Normality Check 2: Q–Q Plot
# -------------------------------------
plt.figure()
stats.probplot(residuals, plot=plt)
plt.title("Normal Q–Q Plot of ANOVA Residuals")
plt.show()
```

9. Assess the constant variance assumption using two methods. Do we have evidence to support the assumption of constant variance? Why or why not?

**Ans**:

## Method 1: Levene’s Test

Levene’s test evaluates whether the response variance is the same across groups:

- **$H_0$:** Variances are equal across treatment combinations  
- **$H_1$:** At least one variance differs  

The test produced **$p > 0.05$**, so we **fail to reject $H_0$**. This means there is **no statistical evidence** that the variances differ across the treatment combinations (i.e., no clear sign of heteroscedasticity).

## Method 2: Residuals vs Fitted Values Plot

The residuals-versus-fitted plot showed:

- Residuals are **randomly scattered around 0**
- **No funnel shape**, curvature, or other systematic pattern
- A **similar spread** of residuals across the range of fitted values  

This visual pattern is consistent with the **constant variance** assumption.

## Conclusion on Constant Variance

Overall, there is **support for the homoscedasticity (constant variance) assumption**. Both Levene’s test and the residual plot suggest that sales variability is **approximately constant** across factor levels, and any small differences in spread are likely due to normal random variation rather than a systematic variance problem.


```python
## Question 9 Code ##
import pandas as pd
import matplotlib.pyplot as plt
import scipy.stats as stats
import statsmodels.api as sm
from statsmodels.formula.api import ols

# -------------------------------------
# Load data
# -------------------------------------
df = pd.read_excel("Busters Burrito Sales.xlsx")

# -------------------------------------
# Fit ANOVA model (main effects only)
# -------------------------------------
model = ols(
    "Sales ~ C(Location) + C(Timing) + C(Marketing)",
    data=df
).fit()

# Extract residuals and fitted values
residuals = model.resid
fitted = model.fittedvalues

# -------------------------------------
# Constant Variance Check 1: Levene's Test
# -------------------------------------
groups = [g["Sales"] for _, g in df.groupby(["Location","Timing","Marketing"])]

lev_test = stats.levene(*groups)

print("\n=== LEVENE'S TEST FOR CONSTANT VARIANCE ===")
print("Test statistic:", lev_test.statistic)
print("p-value:", lev_test.pvalue)

# -------------------------------------
# Constant Variance Check 2: Residuals vs Fitted Plot
# -------------------------------------
plt.figure()
plt.scatter(fitted, residuals)
plt.axhline(0)
plt.xlabel("Fitted Values")
plt.ylabel("Residuals")
plt.title("Residuals vs Fitted Values")
plt.show()
```

10. Report and interpret the ANOVA table. What conclusions can we draw about our main effects? Do the results of these analyses support the null or alternative hypotheses more strongly?

## Interpretation of the ANOVA Table

### Main Effects (α = 0.05)

- **Timing (B)** shows a **statistically significant** effect on sales, meaning average revenue differs between **weekdays and weekends**.
- **Marketing (C)** is also **statistically significant**, with **partnership marketing** producing higher sales than **independent marketing**.
- **Location (A)** appears to have a **smaller effect** and may or may not be statistically significant depending on sampling variability. Even so, the exploratory summaries suggest **brewery locations tend to outperform university locations**.

### Important Note on Aliasing (Defining Relation: *I = ABC*)

Because the design satisfies **$I = ABC$**, each main effect is **aliased with a two-factor interaction**:

- **A** is aliased with **BC**
- **B** is aliased with **AC**
- **C** is aliased with **AB**

So, a statistically significant “main effect” should be interpreted as evidence that **the main effect, the aliased interaction, or both** are contributing to the observed difference. For example, a significant **Timing (B)** effect could reflect **Timing itself**, the **Location × Marketing (AC)** interaction, or a mix of both.

## Hypothesis Support

- The ANOVA provides **stronger support for the alternative hypotheses** for **Timing** and **Marketing**, since these factors show clear mean sales differences.
- Evidence against the null hypothesis for **Location** is **weaker or inconclusive**, although the descriptive results lean toward higher sales at breweries.

```python
## Question 10 Code ##
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Load data
df = pd.read_excel("Busters Burrito Sales.xlsx")

# Fit main-effects ANOVA model
model = ols(
    "Sales ~ C(Location) + C(Timing) + C(Marketing)",
    data=df
).fit()

# ANOVA table (Type II)
anova_table = sm.stats.anova_lm(model, typ=2)

print("\n=== ANOVA TABLE (2^(3-1) FRACTIONAL FACTORIAL) ===")
print(anova_table)

# Significance summary
alpha = 0.05
print("\n=== SIGNIFICANCE SUMMARY (α = 0.05) ===")
for effect in anova_table.index:
    p = anova_table.loc[effect, "PR(>F)"]
    decision = "SIGNIFICANT" if p < alpha else "Not Significant"
    print(f"{effect:15}  p = {p:.4f}  → {decision}")
```

```python
## Question 12 Code ##
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Load data
df = pd.read_excel("Busters Burrito Sales.xlsx")

# Fit main-effects ANOVA model
model = ols(
    "Sales ~ C(Location) + C(Timing) + C(Marketing)",
    data=df
).fit()

anova_table = sm.stats.anova_lm(model, typ=2)
print("\n=== ANOVA TABLE ===")
print(anova_table)

# Residual sum of squares
SSE = anova_table.loc["Residual", "sum_sq"]

# Partial eta-squared
print("\n=== PARTIAL ETA-SQUARED (η²p) ===")
for effect in ["C(Location)", "C(Timing)", "C(Marketing)"]:
    SS = anova_table.loc[effect, "sum_sq"]
    eta_p = SS / (SS + SSE)
    print(f"{effect}: η²p = {eta_p:.3f}")
```

**Ans**:

De-aliasing can be done in a few practical ways, such as running a foldover design, adding the missing runs to complete the full factorial, or switching to a higher-resolution design. That said, de-aliasing isn’t always realistic. It typically requires more experimental runs, which translates directly into more time, higher costs, and greater staffing/operational effort. In real settings (like daily food-truck operations), those added demands can be hard to justify, so teams often accept some aliasing during initial screening and only de-alias if the results are important enough to warrant follow-up testing.
